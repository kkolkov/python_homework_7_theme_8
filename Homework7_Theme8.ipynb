Задание 1. Получение данных из публичного API
Выберите публичный API. Например, JSONPlaceholder. (https://www.google.com/url?q=https%3A%2F%2Fjsonplaceholder.typicode.com%2F)
Напишите скрипт, который: · отправляет GET-запрос к /posts, · извлекает и выводит заголовки и тела первых 5 постов.

```python
try:
    response = requests.get("https://jsonplaceholder.typicode.com/posts")
    response.raise_for_status()  # Проверка статуса
    posts = response.json()
    print("\nПервые 5 постов:\n")
    for post in posts[:5]:
        print(f"ID: {post['id']}")
        print(f"Заголовок: {post['title']}")
        print(f"Тело: {post['body']}\n")
except requests.RequestException as e:
    print("Ошибка при получении данных:", e)
```

Задание 2. Работа с параметрами запроса
Используйте API OpenWeather (https://www.google.com/url?q=https%3A%2F%2Fopenweathermap.org%2Fapi) для получения данных о погоде.
Напишите программу, которая: · принимает название города от пользователя, · отправляет GET-запрос к API и выводит текущую температуру и описание погоды.

```python
API_KEY = "c065b1a4d2fc331b39ac7108353305fe"
city = input("Введите название города: ")

url = "https://api.openweathermap.org/data/2.5/weather"
params = {
    "q": city,
    "appid": API_KEY,
    "units": "metric",  # градусы Цельсия
    "lang": "ru"        # русский язык описания
}

try:
    response = requests.get(url, params=params)
    if response.status_code == 200:
        data = response.json()
        temp = data["main"]["temp"]
        description = data["weather"][0]["description"]
        print(f"\nТемпература в {city}: {temp}°C")
        print(f"Погодные условия: {description}")
    elif response.status_code == 404:
        print("Ошибка 404: город не найден.")
    else:
        print(f"Ошибка {response.status_code}: {response.text}")
except requests.RequestException as e:
    print("Ошибка при обращении к OpenWeather API:", e)
```

Задание 3. Создание и обработка POST-запросов Выберите API, который позволяет создавать ресурсы. Например, JSONPlaceholder. (https://www.google.com/url?q=https%3A%2F%2Fjsonplaceholder.typicode.com%2F) Напишите программу, которая: · отправляет POST-запрос для создания нового поста, · выводит ID созданного поста и его содержимое.

```python
new_post = {
    "title": "Новый пост",
    "body": "Это тестовое сообщение, созданное через API.",
    "userId": 1
}

try:
    response = requests.post("https://jsonplaceholder.typicode.com/posts", json=new_post)
    if response.status_code == 201:
        created_post = response.json()
        print("✅ Пост успешно создан!")
        print("ID:", created_post["id"])
        print("Содержимое:", created_post)
    else:
        print(f"Ошибка {response.status_code}: не удалось создать пост")
except requests.RequestException as e:
    print("Ошибка при создании поста:", e)
```
